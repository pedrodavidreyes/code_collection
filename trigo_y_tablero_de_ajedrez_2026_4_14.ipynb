# Introducción

Este ejemplo está basado en el conocido problema del **trigo y el tablero de ajedrez**, una forma sencilla de observar cómo funciona el crecimiento exponencial.

La idea consiste en colocar **1 grano de trigo en la primera casilla** de un tablero de ajedrez y duplicar la cantidad en cada casilla siguiente: 2 granos en la segunda, 4 en la tercera, 8 en la cuarta, y así sucesivamente hasta completar las 64 casillas del tablero.

Aunque al principio las cantidades parecen pequeñas, el crecimiento se vuelve extremadamente rápido. Este ejercicio utiliza Python para calcular cuántos granos corresponden a cada casilla y cuál es la cantidad total acumulada al finalizar el tablero.

Además de practicar conceptos básicos como listas, ciclos, funciones y operaciones aritméticas, el ejemplo permite visualizar de manera clara el impacto que puede tener una progresión exponencial.


```python
import textwrap
print(textwrap.__file__)
texto = """
Cuenta la leyenda que, cuando se creó el ajedrez, su inventor presentó el juego a un rey. Fascinado por aquella creación, el rey quiso recompensarlo y le ofreció concederle cualquier premio que deseara.

El inventor pidió algo que parecía muy sencillo: colocar 1 grano de trigo en la primera casilla del tablero, 2 en la segunda, 4 en la tercera, y continuar duplicando la cantidad en cada una de las 64 casillas.

El rey consideró que la petición era modesta y fácil de cumplir. Sin embargo, pronto descubriría que aquella sucesión escondía una cantidad extraordinariamente grande.
"""
```

```python
# Este ejemplo está basado en el conocido problema del trigo y el tablero de ajedrez, 
# una forma sencilla de observar cómo funciona el crecimiento exponencial.

# Definimos la función que calcula la cantidad de granos de trigo en cada casilla del tablero de ajedrez
def tablero_ajedrez():
    # Un tablero tiene 8 filas y 8 columnas, un total de 64 casillas
    granos = [1]  # Empieza con 1 grano para la primera casilla
    # la lista granos tiene le valor uno en granos[0], que corresponde a la primera casilla del tablero de ajedrez
    for i in range(0, 63):  # Ciclo  para calcular el valor de las 63 casillas restantes
      granos_siguiente_casilla=granos[i]*2 # Cada casilla tiene el doble de granos que la anterior
      granos.append(granos_siguiente_casilla) # Agrega la cantidad de granos a la lista
    print(granos) # Muestra la cantidad de granos en cada paso del ciclo, es decir, en cada casilla del tablero de ajedrez
    total_de_granos=sum(granos) # Calcula el total de granos sumando todos los elementos de la lista
    print("\nUn total de",total_de_granos,"granos de trigo")
```

```python
print(textwrap.fill(texto, width=80))
print("")
tablero_ajedrez()
print("Nada humilde la petición del inventor del ajedrez, ¿verdad?")
```
